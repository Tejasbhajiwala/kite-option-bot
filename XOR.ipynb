```python
import numpy as np


# Step activation function
def step_function(x):
    if x >= 0:
        return 1
    else:
        return 0


# Generic neuron/perceptron
def neuron(inputs, weights, bias):

    weighted_sum = np.dot(inputs, weights) + bias

    output = step_function(weighted_sum)

    return output


# OR neuron
def OR(x1, x2):

    inputs = np.array([x1, x2])

    weights = np.array([1, 1])

    bias = -0.5

    return neuron(inputs, weights, bias)


# NAND neuron
def NAND(x1, x2):

    inputs = np.array([x1, x2])

    weights = np.array([-1, -1])

    bias = 1.5

    return neuron(inputs, weights, bias)


# XOR neural network
def XOR(x1, x2):

    # Hidden layer
    h1 = OR(x1, x2)

    h2 = NAND(x1, x2)

    # Output neuron = AND
    inputs = np.array([h1, h2])

    weights = np.array([1, 1])

    bias = -1.5

    output = neuron(inputs, weights, bias)

    return output


# Test all XOR combinations
print("X1  X2  XOR")
print("-----------")

for x1, x2 in [(0, 0), (0, 1), (1, 0), (1, 1)]:

    result = XOR(x1, x2)

    print(x1, " ", x2, "  ", result)
```
